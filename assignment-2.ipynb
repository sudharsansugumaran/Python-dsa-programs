```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

def rotate(head, k):
    if head is None or head.next is None:
        return head

    # Find length and last node
    temp = head
    length = 1

    while temp.next:
        temp = temp.next
        length += 1

    k = k % length
    if k == 0:
        return head

    # Make circular
    temp.next = head

    # Find new last node
    for i in range(length - k):
        temp = temp.next

    new_head = temp.next
    temp.next = None

    return new_head

def display(head):
    while head:
        print(head.data, end=" -> ")
        head = head.next
    print("None")

# Create linked list
head = Node(10)
head.next = Node(20)
head.next.next = Node(30)
head.next.next.next = Node(40)
head.next.next.next.next = Node(50)

print("Original List:")
display(head)

head = rotate(head, 2)

print("After Rotation:")
display(head)
```

```python
from collections import deque

d = deque()

# Insert at rear
d.append(10)
d.append(20)

# Insert at front
d.appendleft(5)

print("Deque:", d)

# Delete from front
d.popleft()

# Delete from rear
d.pop()

print("After deletion:", d)
```
